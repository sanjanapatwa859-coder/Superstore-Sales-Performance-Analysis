```python
df.drop_duplicates(inplace = True)
```

```python
df.columns
```

```python
df["Category"] = df["Category"].str.strip()
df["Category"] = df["Category"].str.title()
```

```python
df["Category"] 
```

```python
df.columns = df.columns.str.replace(' ','_')
df.columns = df.columns.str.replace('-','_')
```

```python
df.columns
```

```python
# Empty strings, 'N/A', 'None' ko standard NaN me convert karein
df.replace(['', ' ', 'N/A', 'NA', 'None', 'null'], np.nan, inplace=True)

# Har column me kitne missing values hain dekhein
print(df.isnull().sum())
```

```python
df = df.dropna(subset=['Order_Date'])

df = df.dropna()
```

```python
df['Sales'] = df['Sales'].fillna(df['Sales'].mean())

df['Sales'] = df['Sales'].fillna(df['Sales'].median())
```

```python
mode_value = df['Sub_Category'].mode()[0]
df['Sub_Category'] = df['Sub_Category'].fillna(mode_value)

df['City'] = df['City'].fillna('Unknown')
```

```python
df['Discount'] = df['Discount'].fillna(0)
```

```python
df['Order_Date'] = df['Order_Date'].ffill()

df['Order_Date'] = df['Order_Date'].bfill()
```

```python
df.isnull().sum()
```

```python
df.isnull().values.any()
```

```python
if df.isnull().sum().sum() == 0:
    print(" null ")
else:
    print(" not null")
```

```python
df["Order_Date"].unique()[:10]
```

```python
df["Order_Date"] = df["Order_Date"].str.strip()
```

```python
df['Ship_Date'] = pd.to_datetime(df['Ship_Date'], errors='coerce', format='mixed')
df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors='coerce', format='mixed')

def impute_order_date(row):
    if pd.notnull(row['Order_Date']):
        return row['Order_Date']
    
    ship_mode = str(row['Ship_Mode']).strip()
    days_map = {
        'Same Day': 0,
        'First Class': 2,
        'Second Class': 3,
        'Standard Class': 5
    }
    days = days_map.get(ship_mode, 4)
    return row['Ship_Date'] - pd.Timedelta(days=days)

df['Order_Date'] = df.apply(impute_order_date, axis=1)

df['Year'] = df['Order_Date'].dt.year
df['Month'] = df['Order_Date'].dt.month
df['Quarter'] = df['Order_Date'].dt.quarter

print(df[['Order_Date', 'Year', 'Month', 'Quarter']].head())
```

```python
df["Profit_Margins_Percentage"] = (df["Profit"] / df["Sales"]) * 100
```

```python
df["Discounted_Price"] = df["Sales"] - (df["Sales"] * df["Discount"])
```

```python
df[["Sales" , "Profit" ,"Discount" , "Profit_Margins_Percentage" , "Discounted_Price"]].head(15)
```

```python
monthly_sales = df.groupby("Month")["Sales"].sum()
monthly_sales.plot(kind="line", title="Monthly Sales Trend")
plt.show()
```

```python
plt.figure(figsize = (6,4))
bars = sns.barplot(x="Category", y="Sales", data=df)
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()
```

```python
plt.figure(figsize = (6,4))
sns.scatterplot(data=df, x="Discount",y="Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")
plt.title("Relationship between Discount and Profit")
plt.show()
```

```python
df.to_csv("Superstore_cleaned.csv", index=False)
```
