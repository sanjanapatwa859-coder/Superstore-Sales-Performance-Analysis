```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Superstore_dataset.csv", encoding="latin1")
```

```python
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df.describe(include = 'all')
```

```python
df.isnull().sum()
```

```python
df.drop_duplicates(inplace = True)
```

```python
df["Category"] = df["Category"].str.strip()
df["Category"] = df["Category"].str.title()
```

```python
df["Category"] 
```

```python
df.columns = df.columns.str.replace(' ','_')
```

```python
df.columns = df.columns.str.replace('-','_')
```

```python
df.columns
```

```python
#convert dates
```

```python
df["Order_Date"].unique()[:10]
```

```python
df["Order_Date"] = df["Order_Date"].str.strip()
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"], errors="coerce")
```

```python
df["Year"] = df["Order_Date"].dt.year
```

```python
df["Month"] = df["Order_Date"].dt.month
```

```python
df["Quarter"] = df["Order_Date"].dt.quarter
```

```python
df[["Order_Date","Year","Month","Quarter"]].head()
```

```python
df["Profit_Margins %"] = (df["Profit"] / df["Sales"]) * 100
```

```python
df["Discounted_Price"] = df["Sales"] - (df["Sales"] * df["Discount"])
```

```python
df[["Sales" , "Profit" ,"Discount" , "Profit_Margins %" , "Discounted_Price"]].head()
```

```python
monthly_sales = df.groupby("Month")["Sales"].sum()
monthly_sales.plot(kind="line", title="Monthly Sales Trend")
plt.show()
```

```python
plt.figure(figsize = (6,4))
bars = sns.barplot(x="Category", y="Sales", data=df)
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()
```

```python
plt.figure(figsize = (6,4))
sns.scatterplot(data=df, x="Discount",y="Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")
plt.title("Relationship between Discount and Profit")
plt.show()
```

```python
df.to_csv("Superstore_cleaned.csv", index=False)
```
